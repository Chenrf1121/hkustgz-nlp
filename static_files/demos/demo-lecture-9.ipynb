```python
from collections import defaultdict

# 1. OUR "ROSETTA STONE" (Toy Parallel Corpus)
# Notice how 'casa' appears with 'house' and 'verde' with 'green'.
corpus = [
    (["la", "casa"], ["the", "house"]),
    (["la", "casa", "verde"], ["the", "green", "house"]),
    (["mi", "casa"], ["my", "house"])
]

# 2. INITIALIZE: Every word could translate to any word with equal probability
t = defaultdict(lambda: defaultdict(lambda: 0.25)) 
english_vocab = {"the", "house", "green", "my"}
spanish_vocab = {"la", "casa", "verde", "mi"}

# 3. THE TRAINING LOOP (Simplified EM Algorithm)
for iteration in range(10):
    counts = defaultdict(lambda: defaultdict(float))
    total_e = defaultdict(float)

    for s_sent, e_sent in corpus:
        for e_word in e_sent:
            # E-Step: Compute 'soft' alignment counts
            normalization_factor = sum(t[e_word][s_word] for s_word in s_sent)
            for s_word in s_sent:
                weight = t[e_word][s_word] / normalization_factor
                counts[e_word][s_word] += weight
                total_e[s_word] += weight

    # M-Step: Update probabilities based on counts
    for e_word in english_vocab:
        for s_word in spanish_vocab:
            t[e_word][s_word] = counts[e_word][s_word] / total_e[s_word]

# 4. SHOW RESULTS
print(f"{'English':<10} | {'Spanish':<10} | {'Probability'}")
print("-" * 35)
for e in ["house", "green"]:
    for s in spanish_vocab:
        if t[e][s] > 0.1:
            print(f"{e:<10} | {s:<10} | {t[e][s]:.4f}")
```
